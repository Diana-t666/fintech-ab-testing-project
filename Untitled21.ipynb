```python
import pandas as pd
import numpy as np

# Путь к файлу
file_path = "C:/Users/Di/Downloads/archive (3)/Phonepe-Final-Dataset.xlsx"

# 1. Загружаем только таблицу с транзакциями (первые 5000 строк)
df_transactions = pd.read_excel(file_path, sheet_name='All_Transactions').head(5000).copy()

# 2. Выбираем только нужные колонки для анализа
columns_to_keep = ['Transaction_ID', 'User_ID', 'Amount', 'Service', 'Payment_Status', 'Date']
df_light = df_transactions[columns_to_keep].dropna()

# 3. Делим пользователей на группы А и Б (50/50)
# Берем последнюю цифру из ID: если она четная — группа А, если нечетная — группа Б
df_light['ab_group'] = np.where(df_light['User_ID'].astype(str).str[-1].isin(['0', '2', '4', '6', '8']), 'A', 'B')

# Дополнительно: подкручиваем метрики для группы B (имитация успешного A/B теста)
# Сделаем так, чтобы в группе B суммы транзакций были чуть выше, как будто фича сработала
mask_b = df_light['ab_group'] == 'B'

df_light.loc[mask_b, 'Amount'] = df_light.loc[mask_b, 'Amount'] * 1.07 # +7% к чеку в группе B
df_light['Amount'] = df_light['Amount'].round(2)

# 4. Сохраняем легкий файл для phpMyAdmin
output_file = 'phonepe_light_ab.csv'
df_light.to_csv(output_file, index=False)

print(f"Успешно! Файл '{output_file}' создан.")
print(f"Записано строк: {len(df_light)}")
print("\nРаспределение по группам:")
print(df_light['ab_group'].value_counts())

```

```python
import scipy.stats as stats

# 1. Разделяем суммы успешных транзакций по группам
amount_A = df_light[(df_light['ab_group'] == 'A') & (df_light['Payment_Status'] == 'Successful')]['Amount']
amount_B = df_light[(df_light['ab_group'] == 'B') & (df_light['Payment_Status'] == 'Successful')]['Amount']

# 2. Проводим классический Т-тест Стьюдента для сравнения средних чеков
t_stat, p_value = stats.ttest_ind(amount_A, amount_B, equal_var=False)

print("РЕЗУЛЬТАТЫ СТАТИСТИЧЕСКОГО ТЕСТА:")
print(f"T-статистика: {t_stat:.4f}")
print(f"p-value (уровень значимости): {p_value:.6f}")

# 3. Делаем аналитический вывод на основе p-value
alpha = 0.05
if p_value < alpha:
    print(f"\n Статистически значимый результат! p-value меньше {alpha}.")
    print("Мы отклоняем нулевую гипотезу. Прирост среднего чека в группе B НЕ случаен!")
else:
    print(f"\nРезультат стат-незначим. p-value больше {alpha}.")
    print("Различия между группами могли возникнуть случайно. Внедрять фичу на всех пользователей нельзя.")
```
